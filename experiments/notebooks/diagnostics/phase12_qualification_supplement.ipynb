# Supplemental verification: TED-LIUM qualification caller
This notebook runs the omitted caller suite and checks patch whitespace. It preserves failures and runs no benchmark or final-test inference.

```python
import os, subprocess, sys
from pathlib import Path
ROOT = Path.cwd().resolve()
assert (ROOT / "tests/test_tedlium_qualification.py").is_file()
commands = []
def run(command):
    print("COMMAND:", subprocess.list2cmdline(command), flush=True)
    env = dict(os.environ, PYTHONIOENCODING="utf-8", HF_HUB_OFFLINE="1", TRANSFORMERS_OFFLINE="1")
    result = subprocess.run(command, cwd=ROOT, env=env, capture_output=True, text=True, encoding="utf-8", errors="replace", timeout=600)
    print(result.stdout, end="")
    print(result.stderr, end="", file=sys.stderr)
    print("EXIT:", result.returncode)
    commands.append((command, result.returncode))
run([sys.executable, "-m", "pytest", "-q", "tests/test_tedlium_qualification.py"])
```

```python
run(["git", "diff", "--check"])
assert all(status == 0 for _, status in commands), commands
```
